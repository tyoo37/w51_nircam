```python
import numpy as np
from astropy.table import Table, vstack
from astropy.coordinates import SkyCoord
from astropy import units as u

jwst_complete_catalog = '/orange/adamginsburg/w51/TaehwaYoo/jwst_w51/catalogs/final_catalog_new_cfit_cut.fits'


#W51-E
w51e_alma_catalog = '/blue/adamginsburg/t.yoo/from_red/w51/w51_frag_new/dendro/tables/dendro_w51e_master.fits'
#W51-IRS2
w51n_alma_catalog = '/blue/adamginsburg/t.yoo/from_red/w51/w51_frag_new/dendro/tables/dendro_w51n_master.fits'
print(f'len(w51e_alma_catalog), len(w51n_alma_catalog): {len(Table.read(w51e_alma_catalog))}, {len(Table.read(w51n_alma_catalog))}')
w51e_total_idx = np.arange(len(Table.read(w51e_alma_catalog)))
w51n_total_idx = np.arange(len(Table.read(w51n_alma_catalog)))

w51e_matched_idx = [3, 7, 36, 66, 67, 68, 76, 78, 93, 95, 106]
w51n_matched_idx = [19, 22, 29, 30, 58, 59, 64, 66, 76, 87, 90, 91] #




jwst_catalog_tab = Table.read(jwst_complete_catalog)

skycoord_jwst = jwst_catalog_tab['skycoord_ref']
alma_w51e_tab = Table.read(w51e_alma_catalog[w51e_matched_idx])
alma_w51n_tab = Table.read(w51n_alma_catalog[w51n_matched_idx])
skycoord_w51e_alma = Skycoord(ra=alma_w51e_tab['ra'], dec=alma_w51e_tab['dec'])
skycoord_w51n_alma = Skycoord(ra=alma_w51n_tab['ra'], dec=alma_w51n_tab['dec'])

# skycoord matching
idx_w51e, d2d, d3d = skycoord_w51e_alma.match_to_catalog_sky(skycoord_jwst)
idx_w51n, d2d, d3d = skycoord_w51n_alma.match_to_catalog_sky(skycoord_jwst)

jwst_catalog_tab_matched_w51e = jwst_catalog_tab[idx_w51e]
jwst_catalog_tab_matched_w51n = jwst_catalog_tab[idx_w51n]

jwst_catalog_tab_matched_combined = vstack([jwst_catalog_tab_matched_w51e, jwst_catalog_tab_matched_w51n])

jwst_catalog_tab_unmatched = np.setdiff1d(jwst_catalog_tab, jwst_catalog_tab_matched_combined)
```
