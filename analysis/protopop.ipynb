```python
from protopop.cluster import Cluster
from astropy import units as u
import numpy as np
cl_ca = Cluster(distance=5.4 * u.kpc, timescale=1 * u.Myr, mass=3e2*u.Msun, history='ca', efficiency=100 )

```

```python

sample_time = 0.25*u.Myr
```

```python
sample_ap = np.array([600])*u.AU
sample_wavs = np.array([1.62, 2.10, 3.60, 4.80])*u.um
flux_ca = cl_ca.sample_flux(sample_time, wav=sample_wavs, ap=sample_ap)
```

```python
ev_table = cl_ca.sample_ev(sample_time)
```

```python
print(flux_ca)
print(ev_table)
print(flux_ca[:,0])
```
