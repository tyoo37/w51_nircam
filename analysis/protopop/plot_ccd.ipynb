```python


from astropy.table import Table
from astropy import units as u
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
import numpy as np
from dust_extinction.averages import CT06_MWLoc

def plot_extvec_ccd(ax, color1, color2, ext=CT06_MWLoc(), extvec_scale=30, start=None,
                    color='y', head_width=0.5, color1_data=None, color2_data=None, label=None, fontsize=15):
    def get_wave(c):
        if c == '410m405': return 4.10*u.um
        if c == '405m410': return 4.05*u.um
        if c == 'Hmag': return 1.634*u.um
        if c == 'Ksmag': return 2.143527*u.um
        return int(c[1:-1])/100*u.um

    w1 = get_wave(color1[0])
    w2 = get_wave(color1[1])
    w3 = get_wave(color2[0])
    w4 = get_wave(color2[1])

    if w1 > w2:
        w1, w2 = w2, w1
        color1 = color1[::-1]
    if w3 > w4:
        w3, w4 = w4, w3
        color2 = color2[::-1]
    print(w1,w2,w3,w4)

    # Only plot if all are in valid range
    try:
        e_1 = ext(1/w1) * extvec_scale
        e_2 = ext(1/w2) * extvec_scale
        e_3 = ext(1/w3) * extvec_scale
        e_4 = ext(1/w4) * extvec_scale
    except ValueError as e:
        print(f"Extinction vector not plotted: {e}")
        return
    print(w1,w2,w3,w4)
    print(e_1, e_2, e_3, e_4)
    if start == None:
        mean_color1 = np.nanmean(color1_data)
        mean_color2 = np.nanmean(color2_data)
        std_color1 = np.nanstd(color1_data)
        std_color2 = np.nanstd(color2_data)
        if e_1 - e_2 < 0:
            std_color1 *= -1
        if e_3 - e_4 < 0:
            std_color2 *= -1
    
        start2 = (mean_color1 + 2*std_color1, mean_color2 + 2*std_color2)
    else:
        start2 = start

    ax.annotate(f'$A_V={extvec_scale}$', xy=(start2[0] + (e_1 - e_2), start2[1] + (e_3 - e_4)),
                xytext=(start2[0], start2[1]), fontsize=fontsize,
                arrowprops=dict(arrowstyle='-|>', color=color,
                                shrinkA=0, shrinkB=0,
                                mutation_scale=20, linewidth=1.5))
    
    if label is not None:
        ax.plot([], [], color=color, marker='>', markersize=8,
                label=f'{label}', linestyle='-', linewidth=2)

    else:
        ax.plot([], [], color=color, marker='>', markersize=8,
                linestyle='-', linewidth=2)


savedir = '/orange/adamginsburg/jwst/w51/protopop/'
for history in ['ca', 'is', 'tc', 'expl']: 
    for timescale in ['1.0', '2.0', '4.0']: 
        for efficiency in ['100']: 
            for sample_time in ['0.2', '0.5', '1.0']: 
                try:
                    tab = Table.read(f'{savedir}/ev_table_{history}_mass5000.0_ts_{timescale}_t{sample_time}_eff{efficiency}.fits')
                    tab.pprint(max_width=-1)
                except Exception as e:
                    print(f"Error reading {savedir}/ev_table_{history}_mass5000.0_ts_{timescale}_t{sample_time}_eff{efficiency}.fits: {e}")
                f162m_mag = tab['f162m_mag']
                f210m_mag = tab['f210m_mag']
                f360m_mag = tab['f360m_mag']
                f480m_mag = tab['f480m_mag']

                stellar_mass = tab['Stellar_Mass']
                stage = tab['Stage']
                stellar_luminosity = tab['Total_Luminosity']
                intrinsic_luminosity = tab['Intrinsic_Lum']
                accretion_rate = tab['Accretion_Rate']
                stellar_radius = tab['Stellar_Radius']
                time =tab['Time']
                polytropic_index = tab['Polytropic_Index']


                fig = plt.figure(figsize=(16, 20))
                ax1 = fig.add_subplot(421)
                mass_norm = plt.Normalize(vmin=stellar_mass.min(), vmax=stellar_mass.max())
                cmap = plt.cm.get_cmap('rainbow')
                sc = ax1.scatter(f162m_mag-f210m_mag, f360m_mag-f480m_mag, c=stellar_mass, cmap=cmap, norm=mass_norm, s=3, alpha=0.5)
                cbar = plt.colorbar(sc, label='Stellar Mass (Msun)', orientation='vertical', pad=0.02)
                cbar.set_label('Stellar Mass (Msun)', rotation=270, labelpad=15)
                

                stage_norm = plt.Normalize(vmin=stage.min(), vmax=stage.max())
                cmap_stage = plt.cm.get_cmap('rainbow')
                ax2 = fig.add_subplot(422)
                sc2 = ax2.scatter(f162m_mag-f210m_mag,  f360m_mag-f480m_mag, c=stage, cmap=cmap_stage, norm=stage_norm, s=3, alpha=0.5)
                cbar2 = plt.colorbar(sc2, label='Stage', orientation='vertical', pad=0.02)
                cbar2.set_label('Stage', rotation=270, labelpad=15)
             

                stellar_luminosity_norm = plt.Normalize(vmin=stellar_luminosity.min(), vmax=stellar_luminosity.max())
                ax3 = fig.add_subplot(423)
                sc3 = ax3.scatter(f162m_mag-f210m_mag,  f360m_mag-f480m_mag, c=stellar_luminosity, cmap='rainbow', norm=stellar_luminosity_norm, s=3, alpha=0.5)
                cbar3 = plt.colorbar(sc3, label='Stellar Luminosity (Lsun)', orientation='vertical', pad=0.02)
                cbar3.set_label('Stellar Luminosity (Lsun)', rotation=270, labelpad=15)
           

                stellar_radius_norm = plt.Normalize(vmin=stellar_radius.min(), vmax=stellar_radius.max())
                ax4 = fig.add_subplot(424)
                sc4 = ax4.scatter(f162m_mag-f210m_mag,  f360m_mag-f480m_mag, c=stellar_radius, cmap='rainbow', norm=stellar_radius_norm, s=3, alpha=0.5)
                cbar4 = plt.colorbar(sc4, label='Stellar Radius (Rsun)', orientation='vertical', pad=0.02)
                cbar4.set_label('Stellar Radius (Rsun)', rotation=270, labelpad=15)
               

                accretion_rate_norm = LogNorm(vmin=accretion_rate.min(), vmax=accretion_rate.max())
                ax5 = fig.add_subplot(425)
                sc5 = ax5.scatter(f162m_mag-f210m_mag,  f360m_mag-f480m_mag, c=accretion_rate, cmap='rainbow', norm=accretion_rate_norm, s=3, alpha=0.5)
                cbar5 = plt.colorbar(sc5, label='Accretion Rate (Msun/yr)', orientation='vertical', pad=0.02)
                cbar5.set_label('Accretion Rate (Msun/yr)', rotation=270, labelpad=15)
          

                intrinsic_luminosity_norm = LogNorm(vmin=np.nanmin(intrinsic_luminosity), vmax=np.nanmax(intrinsic_luminosity))
                ax6 = fig.add_subplot(426)

                good = np.isfinite(intrinsic_luminosity) & (intrinsic_luminosity > 0)

                il = intrinsic_luminosity[good]
                x = f162m_mag[good] - f210m_mag[good]
                y = f360m_mag[good] - f480m_mag[good]

                vmin = np.nanmin(il)
                vmax = np.nanmax(il)

                # Guard against degenerate case where all values are identical
                if vmax <= vmin:
                    vmax = vmin * 1.0001

                intrinsic_luminosity_norm = LogNorm(vmin=vmin, vmax=vmax)
                sc6 = ax6.scatter(x, y, c=il, cmap='rainbow',
                                    norm=intrinsic_luminosity_norm, s=3, alpha=0.5)
                cbar6 = plt.colorbar(sc6, label='Intrinsic Luminosity (Lsun)',
                                        orientation='vertical', pad=0.02)
                cbar6.set_label('Intrinsic Luminosity (Lsun)', rotation=270, labelpad=15)
             

                ax7 = fig.add_subplot(427)
                time_norm = plt.Normalize(vmin=time.min(), vmax=time.max())
                sc7 = ax7.scatter(f162m_mag-f210m_mag,  f360m_mag-f480m_mag, c=time, cmap='rainbow', norm=time_norm, s=3, alpha=0.5)
                cbar7 = plt.colorbar(sc7, label='Time (Myr)', orientation='vertical', pad=0.02)
                cbar7.set_label('Time (Myr)', rotation=270, labelpad=15)
       

                ax8 = fig.add_subplot(428)
                polytropic_index_norm = plt.Normalize(vmin=polytropic_index.min(), vmax=polytropic_index.max())
                sc8 = ax8.scatter(f162m_mag-f210m_mag,  f360m_mag-f480m_mag, c=polytropic_index, cmap='rainbow', norm=polytropic_index_norm, s=3, alpha=0.5)
                cbar8 = plt.colorbar(sc8, label='Polytropic Index', orientation='vertical', pad=0.02)
                cbar8.set_label('Polytropic Index', rotation=270, labelpad=15)
                ext_model = CT06_MWLoc()
                ax1.set_title(f'Mass: 5e3Msun, History: {history}, Timescale: {timescale} Myr, Sample Time: {sample_time} Myr, Eff: {efficiency}%')
                for ax in [ax1, ax2, ax3, ax4, ax5, ax6, ax7, ax8]:

                    plot_extvec_ccd(ax, color1=('f162m', 'f210m'), color2=('f360m', 'f480m'),
                        ext=ext_model, extvec_scale=30, color='magenta',
                        start=(0,2))
                    ax.set_xlabel('F162M - F210M (mag)')
                    ax.set_ylabel('F360M - F480M (mag)')
                   
                    ax.set_xlim(-1,  4)
                    ax.set_ylim(-1,  4)

                plt.savefig(f'{savedir}/ccd_protopop_{history}_mass5000.0_ts_{timescale}_t{sample_time}_eff{efficiency}.png', dpi=300, bbox_inches='tight')
                plt.show()
```
