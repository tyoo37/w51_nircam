```python
import numpy as np
import matplotlib.pyplot as plt
names = ['W51b1', 'W51b2', 'W51e7']
logQ0 = [48.61, 48.02, 47.71]
diameter_arcsec =  [30, 7, 5.4]
diameter_pc = np.array(diameter_arcsec) * 5400 / 206265



fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111)
ax.scatter(np.log10(diameter_pc), logQ0, s=100)
for i in range(len(names)):
    ax.annotate(names[i], (np.log10(diameter_pc[i]), logQ0[i]), xytext=(5, 5), textcoords='offset points')
        
ax.set_xlabel('Diameter (pc)')
ax.set_ylabel('log Q0')
ax.set_ylim(44,50)
ax.set_xlim(-2,0)
plt.show()
```
