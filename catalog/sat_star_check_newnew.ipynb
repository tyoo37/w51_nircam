```python
import glob
from astropy.io import fits
import matplotlib.pyplot as plt
from astropy.visualization import simple_norm
from astropy.table import Table
from jwst.datamodels import dqflags
from scipy.ndimage import label, find_objects, center_of_mass, sum_labels
import numpy as np
filt = 'F480M'
for module in ('nrca', 'nrcb'):
    for fn in glob.glob(f"/orange/adamginsburg/jwst/w51/{filt}/pipeline/*{module}*align*crf.fits"):
        satstar_fn = fn.replace('.fits', '_satstar_catalog_new.fits')
        print(fn)
        print(satstar_fn)
        img = fits.open(fn)['SCI'].data
        header = fits.open(fn)[0].header
        fig = plt.figure(figsize=(20,20))
        ax1 = fig.add_subplot(1,2,1)
        ax2 = fig.add_subplot(1,2,2)
        norm = simple_norm(img, stretch='log', min_percent=99.5)
        ax1.imshow(img, norm=norm, origin='lower', cmap='inferno', interpolation='nearest')
        dq = fits.open(fn)['DQ'].data
        mask = (dq & dqflags.group['SATURATED'])>0
        ax2.imshow(mask, origin='lower', cmap='gray', interpolation='nearest')

        
        satstar_tab = Table.read(satstar_fn)
        x = satstar_tab['x_fit']
        y = satstar_tab['y_fit']
        ax1.scatter(x, y, color='cyan', s=100, edgecolor='cyan', facecolor='none', lw=2, label='Saturated Stars')
        ax2.scatter(x, y, color='cyan', s=100, edgecolor='cyan', facecolor='none', lw=2, label='Saturated Stars')
        plt.show()
        sources_, nsources_ = label(mask)
        slices = find_objects(mask)
       
        pad=30
        plt.show()
        plt.close()
     

        for i in range(nsources_-1):
            # get the center of pixels with this label
            com = center_of_mass(mask, labels=sources_, index=i+1)
            ycen, xcen = map(int, map(round, com))
            print(f"Source {i+1}: center at (x, y) = ({xcen}, {ycen})")
            y0 = max(0, ycen - pad)
            y1 = min(img.shape[0], ycen + pad)
            x0 = max(0, xcen - pad)
            x1 = min(img.shape[1], xcen + pad)
            cutout = img[y0:y1, x0:x1]
            fig = plt.figure(figsize=(20,10))
            ax1 = fig.add_subplot(1,2,1)
            ax1.imshow(cutout, norm=norm, origin='lower', cmap='inferno', interpolation='nearest')
            ax1.scatter(x-x0, y-y0, color='cyan', s=100, edgecolor='cyan', facecolor='none', lw=2)
            ax1.set_xlim(0, cutout.shape[1])
            ax1.set_ylim(0, cutout.shape[0])
            ax2 = fig.add_subplot(1,2,2)
            cutout_mask = mask[y0:y1, x0:x1]
            ax2.imshow(cutout_mask, origin='lower')
            plt.show()
            plt.close()
                
```
