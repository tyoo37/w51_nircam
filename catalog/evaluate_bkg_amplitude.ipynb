```python
#To estimate how good or bad this background estimate is, we can operate on a star-masked-out image and measure the amplitude of the gradient on 0.9” = 0.03”pixscale * 30 pixels and 1.8” = 0.06”pixscale * 30 pixels scales.  
#E.g., smooth data (w/o stars) to 0.9 or 1.8” resolution, then take the diff along the X and Y axes, square that, then sum it: [(data[:-1,:-1] - data[1:,:-1])**2 + (data[:-1,:-1] - data[:-1,1:])**2]**0.5

import numpy as np
from astropy.table import Table, vstack
from astropy.coordinates import SkyCoord


def get_bkg_gradient(bkg_image):
    """Estimate background gradient amplitude in image.

    Parameters
    ----------
    bkg_image : 2D array
        Background image.

    Returns
    -------
    gradient_amplitude : float
        Estimated background gradient amplitude.
    """
    dy, dx = np.gradient(bkg_image)
    gradient_amplitude = np.sqrt(dx**2 + dy**2)
    return np.sum(gradient_amplitude)
```
