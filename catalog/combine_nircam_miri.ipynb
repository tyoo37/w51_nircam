```python
from astropy import units as u
import numpy as np
nircam_catalog = '/orange/adamginsburg/jwst/w51/catalogs/final_nircam_indivexp_merged_dao_refined_after_sat.fits'
miri_catalog = '/orange/adamginsburg/jwst/w51/catalogs/final_miri_indivexp_merged_dao_refined_after_sat.fits'

# combine two catalog based on the skycoord
from astropy.table import Table, vstack
nircam_cat = Table.read(nircam_catalog)
miri_cat = Table.read(miri_catalog)

print(nircam_cat.colnames)
nircam_skycoord = nircam_cat['skycoord']
miri_skycoord = miri_cat['skycoord']
idx, d2d, d3d = miri_skycoord.match_to_catalog_sky(nircam_skycoord)
matched = d2d < 0.1*u.arcsec
miri_keep = np.where(matched==False)[0]
combined_cat = vstack([nircam_cat, miri_cat[miri_keep]])
print(f'Number of sources in NIRCam catalog: {len(nircam_cat)}')
print(f'Number of sources in MIRI catalog: {len(miri_cat)}')
print(f'Number of matched sources between NIRCam and MIRI within 0.1 arcsec: {len(miri_cat)-len(miri_keep)}')
print(f'Number of sources in the combined catalog: {len(combined_cat)}')    

combined_cat.write('/orange/adamginsburg/jwst/w51/catalogs/final_nircam_miri_indivexp_merged_dao_refined_after_sat.fits', overwrite=True)
```
