```python
jwst_matched_ra = np.ones(len(tab_gtc['ra_deg']))*np.nan
jwst_matched_dec = np.ones(len(tab_gtc['ra_deg']))*np.nan
is_jwstmatched = np.zeros(len(tab_gtc['ra_deg']), dtype=bool)
close_neighbor = np.zeros(len(tab_gtc['ra_deg']), dtype=bool)
for i in range(len(tab_gtc['ra_deg'])):
    if i in list(manual_matching.keys()):
        if manual_matching[i]['module'] == 'nrca':
            catalog = nrca_catalog
        else:
            catalog = nrcb_catalog
        source_id = manual_matching[i]['id']
        jwst_matched_skycoord = catalog['skycoord_ref'][catalog['id_f140m']==source_id][0]
        jwst_matched_ra[i] = jwst_matched_skycoord.ra.deg
        jwst_matched_dec[i] = jwst_matched_skycoord.dec.deg
        is_jwstmatched[i] = True
        close_neighbor[i] = manual_matching[i]['close_neighbor']
    elif i in list(manual_matching_sat.keys()):
        ra = manual_matching_sat[i]['ra']
        dec = manual_matching_sat[i]['dec']
        jwst_matched_skycoord = SkyCoord(ra=ra, dec=dec, frame='icrs', unit=(u.hourangle, u.deg))
        jwst_matched_ra[i] = jwst_matched_skycoord.ra.deg
        jwst_matched_dec[i] = jwst_matched_skycoord.dec.deg
        is_jwstmatched[i] = True
        close_neighbor[i] = manual_matching_sat[i]['close_neighbor']



```

```python
from regions import Regions, CircleSkyRegion, CirclePixelRegion
from astropy.coordinates import SkyCoord

import re

region_file = "/home/t.yoo/w51/w51_jwst_sources_highlight/regions/gtc_spec_target3.reg"

pattern = re.compile(
    r"point\(([^,]+),\s*([^)]+)\)\s*#.*?text=\{([^}]+)\}"
)

labels = []
with open(region_file) as f:
    for line in f:
        match = pattern.search(line)
        if match:
            ra = float(match.group(1))
            dec = float(match.group(2))
            label = match.group(3)
            labels.append(label)



fig = plt.figure(figsize=(24, 24))
ax1 = fig.add_subplot(111, projection=WCS(f480_header, naxis=2))
f480_img = fits.open(image_filenames['f480m'])[1].data
f480_header = fits.open(image_filenames['f480m'])[1].header
ax1.imshow(f480_img, cmap='gray', origin='lower', norm=simple_norm(f480_img, stretch='log',percent=99.9))
jwst_matched_skycoord = SkyCoord(ra=jwst_matched_ra*u.deg, dec=jwst_matched_dec*u.deg, frame='icrs')
jwst_matched_pixcoord = jwst_matched_skycoord.to_pixel(wcs=WCS(f480_header, naxis=2))
ax1.scatter(jwst_matched_pixcoord[0][is_jwstmatched], jwst_matched_pixcoord[1][is_jwstmatched], color='yellow', s=5)
regions = Regions.read('/home/t.yoo/w51/w51_jwst_sources_highlight/regions/gtc_spec_target3.reg', format='ds9')
wcs_f140m = WCS(f480_header, naxis=2)
reg_name_col = np.full(len(tab_gtc['ra_deg']), '', dtype='U20')
jwst_matched_skycoord2 = SkyCoord(list(jwst_matched_ra), list(jwst_matched_dec), unit='deg', frame='icrs')
print(jwst_matched_skycoord2)
for coord in jwst_matched_skycoord2:
    print(coord)
for i, reg in enumerate(regions):
    print(reg)
    does_contain = np.array([CircleSkyRegion(center=reg.center, radius=0.1*u.arcsec).contains(coord, wcs=wcs_f140m) for coord in jwst_matched_skycoord2])
    reg_expand = CircleSkyRegion(center=reg.center, radius=0.1*u.arcsec)
    if any(does_contain):
        print('ho')
        reg_name_col[does_contain] = labels[i]
    pixel_region = reg_expand.to_pixel(wcs=wcs_f140m)
    pixel_region.plot(ax=ax1, color='red', lw=1, alpha=0.5)

print(reg_name_col)


```
